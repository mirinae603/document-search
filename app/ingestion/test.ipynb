```python
import os
import asyncio
from extractor import DocumentExtractor

def get_mime_type(file_path: str) -> str:
    ext = os.path.splitext(file_path)[-1].lower()
    return {
        ".pdf":  "application/pdf",
        ".docx": "application/vnd.openxmlformats-officedocument.wordprocessingml.document",
        ".pptx": "application/vnd.openxmlformats-officedocument.presentationml.presentation",
        ".xlsx": "application/vnd.openxmlformats-officedocument.spreadsheetml.sheet",
        ".csv":  "text/csv",
        ".txt":  "text/plain",
        ".md":   "text/markdown",
        ".html": "text/html",
        ".jpg":  "image/jpeg",
        ".png":  "image/png",
    }.get(ext, "application/octet-stream")

async def test_extraction(file_path: str):
    print(f"\n📄 Reading: {file_path}")

    if not os.path.exists(file_path):
        print("❌ File not found")
        return

    with open(file_path, "rb") as f:
        content = f.read()

    mime_type = get_mime_type(file_path)
    print(f"📌 MIME: {mime_type}")

    extractor = DocumentExtractor()

    print("\n🚀 Extracting...\n")

    try:
        pages = await extractor.extract_pages(content, mime_type)
    except Exception as e:
        print(f"❌ Crash: {e}")
        return

    if not pages:
        print("⚠️ No pages extracted")
        return

    print(f"\n✅ Total Pages: {len(pages)}\n")

    for page in pages:
        print("\n" + "="*50)
        print(f"📄 PAGE {page['page_num']}")
        print("="*50)

        # Print the EXACT raw output from the extractor
        raw = page.get("text", "")
        
        print(raw[:15000])  # preview

        if len(raw) > 15000:
            print("\n... (truncated)")

    return pages


# --- RUN ---
# If running in a Jupyter Notebook:
await test_extraction("testing_3.pdf")

# If running as a standard Python script (.py file in terminal), uncomment this instead:
# if __name__ == "__main__":
#     asyncio.run(test_extraction("testing_1.pdf"))
```
